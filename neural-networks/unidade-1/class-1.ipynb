```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def step(x):
    """
    Args:
        x : value for step function
    Return:
        if x >= 0 return 1
        else     return 0
    """
    if x >= 0:
        return 1
    return 0


def neuron(X, W, theta):
    """

    Args:
        X : input array
        W : weight array
        theta: bias
    Return:
        y : output of neuron
    """
    soma = np.dot(X, W) + theta

    y = step(soma)

    return y
```

```python
X = np.array([1, 0])
W = np.array([0.5, 0.2])
theta = -0.5

output = neuron(X, W, theta)
print(output)
```

```python
class Perceptron:
    def __init__(self, input_size):
        self.weights = np.zeros(input_size)
        self.bias = 0

    def output(self, inputs):
        sum = np.dot(inputs, self.weights) + self.bias

        return self.step_function(sum)

    def step_function(self, x):
        if x >= 0:
            return 1
        return 0

    def train(self, input, output, epoch=10, learning_rate = 1):
        for epoch in range(epoch):
            for input_in, output_in in zip(input, output):
                output_perceptron = self.output(input_in)
                self.weights += learning_rate * (output_in - output_perceptron) * input_in
                self.bias +=  learning_rate * (output_in - output_perceptron)

            print(f" Epoch: {epoch+1:>02}", " Error: ", (output_in - output_perceptron))
```

## Porta OR

```python
inputs = np.array([[0,0], [0, 1], [1,0], [1,1]])
outputs = np.array([0, 1, 1, 1])

perceptron = Perceptron(2) # 2 entradas por vez

perceptron.train(inputs, outputs, epoch=15, learning_rate=0.1)
```

```python
def plot_hyperplane(weights, bias):
    input = np.linspace(-0.1, 1.1, 100)

    h = (-bias - weights[0] * input)/weights[1]  # equação do hiperplano
    plt.plot(input, h, label="Hiperplano encontrado")
```

```python
plot_hyperplane(perceptron.weights, perceptron.bias)
plt.scatter(inputs[:, 0], inputs[:,1], color='r')
plt.xlabel("Entrada 1")
plt.ylabel("Entrada 2")
plt.legend()
plt.show()
```
